**WEB SCRAPING**

Write a Python program that extracts information from a webpage using Beautiful Soup. The program should retrieve the titles and release years of the top 100 movies from the IMDb Top Rated Movies page (https://www.imdb.com/chart/top).

**Step 1: Import the necessary libraries**

```python
# Step 1: import the libraries we will use
# requests  -> downloads the web page from the internet
# bs4       -> contains BeautifulSoup, which helps us parse (read) the HTML
import requests
from bs4 import BeautifulSoup
```

**Step 2: Send a GET request to the IMDb Top Rated Movies page**

```python
# Step 2: send a GET request to the IMDb Top Rated Movies page
url = "https://www.imdb.com/chart/top/"  # web address of the Top Rated Movies chart

# requests.get() asks the server for the page
response = requests.get(url)

# Save the HTML content (as plain text) so that BeautifulSoup can read it later
html_content = response.text

# Print the status code so we can quickly see if the request worked (200 means OK)
print("Status code:", response.status_code)
```

**Step 3: Parse the HTML content using Beautiful Soup**

```python
# Step 3: parse the HTML using BeautifulSoup
# BeautifulSoup builds a tree structure from the raw HTML text so we can search it easily.
soup = BeautifulSoup(html_content, "html.parser")

# Just to check that parsing worked, we can print the title of the page.
print("Page title:", soup.title.string.strip())
```

**Step 4: Find and extract the titles and release years of the top 10 movies**

```python
# Step 4: find the table rows that contain movie information
# On the IMDb Top Rated Movies page, each movie is one <tr> (table row)
# inside <tbody class="lister-list">.
movie_rows = soup.select("tbody.lister-list tr")

print("Total movies found on the page:", len(movie_rows))
```

**Step 5: Store the extracted information**

```python
# Step 5: store the extracted information
# We will build a list of dictionaries. Each dictionary holds
# the title and year for one movie.

top_100_movies = []  # this will store information for the top 100 movies

for row in movie_rows[:100]:  # take only the first 100 rows (top 100 movies)
    # Get the movie title from the <a> tag in the "titleColumn" cell
    title_tag = row.select_one("td.titleColumn a")
    title = title_tag.get_text(strip=True)

    # Get the release year from the <span class="secondaryInfo"> tag
    year_tag = row.select_one("td.titleColumn span.secondaryInfo")
    year = year_tag.get_text(strip=True).strip("()")  # remove surrounding parentheses

    # Store the pieces of information in a small dictionary
    movie_info = {"title": title, "year": year}
    top_100_movies.append(movie_info)

# The top 10 movies are simply the first 10 of the top_100_movies list
top_10_movies = top_100_movies[:10]

print("Number of movies stored (should be 100 or less):", len(top_100_movies))
```

**Step 6: Display the extracted movie information**

```python
# Step 6: display the extracted movie information
# Here we print only the top 10 movies (title and year) to keep the
# output short and easy to read. The full top 100 list is still stored
# in the top_100_movies variable.

print("Top 10 movies (title and year):\n")
for movie in top_10_movies:
    print(f"{movie['title']} ({movie['year']})")
```

**QUESTION 2: Can you modify the code to extract the movie ratings as well? Display the top 10 movies with their ratings.**

```python
# QUESTION 2: include the IMDb rating for each of the top 10 movies
# We will rebuild a list of the top 10 movies, but this time we also
# read the rating from the "imdbRating" column.

top_10_with_ratings = []

for row in movie_rows[:10]:  # still looking at the top 10 rows
    # Title and year (same as before)
    title = row.select_one("td.titleColumn a").get_text(strip=True)
    year_text = row.select_one("td.titleColumn span.secondaryInfo").get_text(strip=True)
    year = year_text.strip("()")

    # The rating value is stored inside <strong> under the imdbRating column
    rating_text = row.select_one("td.imdbRating strong").get_text(strip=True)
    rating = float(rating_text)  # convert the rating text to a float number

    top_10_with_ratings.append({"title": title, "year": year, "rating": rating})

print("Top 10 movies with ratings:\n")
for movie in top_10_with_ratings:
    print(f"{movie['title']} ({movie['year']}): rating = {movie['rating']}")
```

**QUESTION 3: Can you filter the list of movies which have rating greater than 9**

```python
# QUESTION 3: filter the movies so that we only keep those with rating > 9
# We already have a list called top_10_with_ratings. We can use a
# list comprehension to build a new filtered list.

highly_rated_movies = [movie for movie in top_10_with_ratings if movie["rating"] > 9]

print("Movies with rating > 9 (from the top 10 list):\n")
for movie in highly_rated_movies:
    print(f"{movie['title']} ({movie['year']}): rating = {movie['rating']}")
```
